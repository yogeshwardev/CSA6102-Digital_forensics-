```python
baseline = [
    ("admin_log.txt", 2),
    ("bash_history.txt", 4),
    ("sysytem_logs.txt", 6),
    ("windows_process.txt", 0)
]

system = [
    ("admin_log.txt", 2),
    ("bash_history.txt", 4),
    ("sysytem_logs.txt", 4),
    ("windows_process.txt", 6)
]

for files, acc in baseline:
    for file, acce in system:
        if files == file and acc != acce:
            print(files, acc, acce)
```
