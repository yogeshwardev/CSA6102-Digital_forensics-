```python
## import hashlib
from datetime import datetime
from collections import defaultdict

# ============================================================
# CAMPUS DIGITAL FORENSICS EVIDENCE MANAGEMENT
# AND INCIDENT RECONSTRUCTION SYSTEM
# STATIC DEMONSTRATION VERSION
# ============================================================

CASE_ID = "CASE-2026-0311"
WORKSTATION = "WKSTN-14"
LEAD_INVESTIGATOR = "A. Kumar"

evidence_register = []
custody_log = []


# ============================================================
# STATIC SIMULATED EVIDENCE
# ============================================================

EVIDENCE_CONTENT = {
    "EVD-001":
        b"Simulated forensic disk image of WKSTN-14",

    "EVD-002":
        b"Simulated RAM capture of WKSTN-14",

    "EVD-003":
        b"Simulated domain controller authentication logs",

    "EVD-004":
        b"Simulated perimeter firewall network logs"
}


EVIDENCE_DATA = [
    {
        "evidence_id": "EVD-001",
        "source": "WKSTN-14 Hard Disk Image",
        "acquisition": "2026-03-11 09:15:00",
        "investigator": "A. Kumar",
        "storage": "Locker-3 / Shelf-B"
    },
    {
        "evidence_id": "EVD-002",
        "source": "WKSTN-14 RAM Capture",
        "acquisition": "2026-03-11 09:22:00",
        "investigator": "A. Kumar",
        "storage": "Locker-3 / Shelf-B"
    },
    {
        "evidence_id": "EVD-003",
        "source": "Domain Controller Auth Log Export",
        "acquisition": "2026-03-11 10:05:00",
        "investigator": "S. Rao",
        "storage": "Locker-3 / Shelf-C"
    },
    {
        "evidence_id": "EVD-004",
        "source": "Perimeter Firewall Network Log Export",
        "acquisition": "2026-03-11 10:20:00",
        "investigator": "S. Rao",
        "storage": "Locker-3 / Shelf-C"
    }
]


# ============================================================
# STATIC CHAIN OF CUSTODY
# ============================================================

CUSTODY_DATA = [
    {
        "evidence_id": "EVD-001",
        "handler": "A. Kumar",
        "timestamp": "2026-03-11 09:15:00",
        "action": "Collected",
        "purpose": "Disk image acquired using write blocker"
    },
    {
        "evidence_id": "EVD-001",
        "handler": "A. Kumar",
        "timestamp": "2026-03-11 09:40:00",
        "action": "Transferred",
        "purpose": "Moved to Locker-3 / Shelf-B"
    },
    {
        "evidence_id": "EVD-001",
        "handler": "S. Rao",
        "timestamp": "2026-03-12 14:00:00",
        "action": "Checked Out",
        "purpose": "Static forensic examination"
    },
    {
        "evidence_id": "EVD-001",
        "handler": "S. Rao",
        "timestamp": "2026-03-12 17:30:00",
        "action": "Returned",
        "purpose": "Examination completed"
    },
    {
        "evidence_id": "EVD-002",
        "handler": "A. Kumar",
        "timestamp": "2026-03-11 09:22:00",
        "action": "Collected",
        "purpose": "Live RAM capture before shutdown"
    },
    {
        "evidence_id": "EVD-003",
        "handler": "S. Rao",
        "timestamp": "2026-03-11 10:05:00",
        "action": "Collected",
        "purpose": "Authentication log export"
    },
    {
        "evidence_id": "EVD-004",
        "handler": "S. Rao",
        "timestamp": "2026-03-11 10:20:00",
        "action": "Collected",
        "purpose": "Firewall network log export"
    }
]


# ============================================================
# STATIC BASELINE
# ============================================================

BASELINE = {
    "payroll_master.xlsx":
        "3a5512dbb69f1e4f79c620441827301e909f363e57446b1c0999c0550d59bba0",

    "hr_policies.docx":
        "ade5e5b40501fcb8645afb99043074cc32273df09fea2f4086ae18084bbe3cd1",

    "finance_report_q1.pdf":
        "3de3267d4c0e59e89088ee5de8aafa4757ce9226ca79df81293f9588d7c3517a",

    "system_config.ini":
        "62816d25428ba1279d550decf29d97e78c6321cc5a89b965df6a27e81ae31c5d",

    "backup_schedule.txt":
        "074da2dfc20fe8549d12f05a0e964b41a624b5a925938c30a7d895726b1904bb",

    # Demonstration case for rename detection
    "security_policy_old.txt":
        "1111111111111111111111111111111111111111111111111111111111111111"
}


CURRENT_SNAPSHOT = {
    "payroll_master.xlsx":
        "bce359fb8ff9f644f0d9911469ef2732e2f7510df3e5f3732b82cbbdb7bf067c",

    "hr_policies.docx":
        "ade5e5b40501fcb8645afb99043074cc32273df09fea2f4086ae18084bbe3cd1",

    "system_config.ini":
        "62816d25428ba1279d550decf29d97e78c6321cc5a89b965df6a27e81ae31c5d",

    "backup_schedule.txt":
        "074da2dfc20fe8549d12f05a0e964b41a624b5a925938c30a7d895726b1904bb",

    "svc_update.exe":
        "bd3826c70cb8dd991e7a286f558dd95815b2a3199109605f27c6f462671f5eeb",

    # Same hash = renamed file
    "security_policy_new.txt":
        "1111111111111111111111111111111111111111111111111111111111111111"
}


# ============================================================
# STATIC AUTHENTICATION LOGS
# ============================================================

AUTH_LOGS = [
    ("2026-03-10 23:58:02", "jsmith", "203.0.113.44",
     "FAILED_LOGIN", "WKSTN-14"),

    ("2026-03-10 23:58:11", "jsmith", "203.0.113.44",
     "FAILED_LOGIN", "WKSTN-14"),

    ("2026-03-10 23:58:19", "jsmith", "203.0.113.44",
     "FAILED_LOGIN", "WKSTN-14"),

    ("2026-03-10 23:58:27", "jsmith", "203.0.113.44",
     "FAILED_LOGIN", "WKSTN-14"),

    ("2026-03-10 23:58:35", "jsmith", "203.0.113.44",
     "FAILED_LOGIN", "WKSTN-14"),

    ("2026-03-10 23:58:44", "jsmith", "203.0.113.44",
     "FAILED_LOGIN", "WKSTN-14"),

    ("2026-03-10 23:58:52", "jsmith", "203.0.113.44",
     "SUCCESSFUL_LOGIN", "WKSTN-14"),

    ("2026-03-11 00:15:30", "jsmith", "203.0.113.44",
     "SUCCESSFUL_LOGIN", "WKSTN-14"),

    ("2026-03-11 07:45:00", "jsmith", "10.0.4.22",
     "SUCCESSFUL_LOGIN", "WKSTN-14")
]


# ============================================================
# STATIC FILE ACTIVITY
# ============================================================

FILE_LOGS = [
    (
        "2026-03-10 23:59:10",
        "WKSTN-14",
        "jsmith",
        "CREATE",
        "svc_update.exe"
    ),

    (
        "2026-03-11 00:05:20",
        "WKSTN-14",
        "jsmith",
        "MODIFY",
        "payroll_master.xlsx"
    ),

    (
        "2026-03-11 00:08:45",
        "WKSTN-14",
        "jsmith",
        "DELETE",
        "finance_report_q1.pdf"
    ),

    (
        "2026-03-11 00:10:10",
        "WKSTN-14",
        "jsmith",
        "RENAME",
        "security_policy_old.txt -> security_policy_new.txt"
    )
]


# ============================================================
# STATIC NETWORK LOGS
# ============================================================

NETWORK_LOGS = [
    (
        "2026-03-11 00:22:05",
        "WKSTN-14",
        "198.51.100.77",
        4444,
        15820000,
        "TCP"
    ),

    (
        "2026-03-11 00:23:40",
        "WKSTN-14",
        "198.51.100.77",
        4444,
        9400000,
        "TCP"
    ),

    (
        "2026-03-11 08:05:00",
        "WKSTN-14",
        "10.0.1.5",
        443,
        52000,
        "TCP"
    )
]


# ============================================================
# HASH FUNCTION
# ============================================================

def calculate_sha256(content):
    return hashlib.sha256(content).hexdigest()


# ============================================================
# 1. EVIDENCE REGISTRATION
# ============================================================

def register_evidence():

    evidence_register.clear()

    print("\n" + "=" * 100)
    print("1. EVIDENCE REGISTRATION AND SHA-256 HASHING")
    print("=" * 100)

    for evidence in EVIDENCE_DATA:

        evidence_id = evidence["evidence_id"]

        sha256 = calculate_sha256(
            EVIDENCE_CONTENT[evidence_id]
        )

        record = evidence.copy()
        record["sha256"] = sha256

        evidence_register.append(record)

        print("\nEvidence ID      :", evidence_id)
        print("Source           :", evidence["source"])
        print("Acquisition Time :", evidence["acquisition"])
        print("Investigator     :", evidence["investigator"])
        print("Storage Location :", evidence["storage"])
        print("SHA-256          :", sha256)

    print("\nSTATUS: All evidence registered successfully.")


# ============================================================
# 2. ADD CHAIN OF CUSTODY
# ============================================================

def load_chain_of_custody():

    custody_log.clear()

    for entry in CUSTODY_DATA:
        custody_log.append(entry.copy())

    print("\n" + "=" * 100)
    print("2. CHAIN-OF-CUSTODY ENTRIES")
    print("=" * 100)

    print(
        "\nAll simulated custody entries have been "
        "loaded successfully."
    )

    print(
        "Total custody events:",
        len(custody_log)
    )


# ============================================================
# 3. DISPLAY CHAIN OF CUSTODY
# ============================================================

def display_chain_of_custody():

    if not custody_log:
        load_chain_of_custody()

    print("\n" + "=" * 115)
    print("3. CHAIN-OF-CUSTODY HISTORY")
    print("=" * 115)

    print(
        f"{'Evidence':<12}"
        f"{'Timestamp':<22}"
        f"{'Handler':<15}"
        f"{'Action':<18}"
        f"{'Purpose'}"
    )

    print("-" * 115)

    for entry in custody_log:

        print(
            f"{entry['evidence_id']:<12}"
            f"{entry['timestamp']:<22}"
            f"{entry['handler']:<15}"
            f"{entry['action']:<18}"
            f"{entry['purpose']}"
        )


# ============================================================
# 4. INTEGRITY VERIFICATION
# ============================================================

def verify_integrity():

    if not evidence_register:
        register_evidence()

    print("\n" + "=" * 100)
    print("4. EVIDENCE INTEGRITY VERIFICATION")
    print("=" * 100)

    for record in evidence_register:

        evidence_id = record["evidence_id"]

        recorded_hash = record["sha256"]

        current_hash = calculate_sha256(
            EVIDENCE_CONTENT[evidence_id]
        )

        status = (
            "MATCH"
            if recorded_hash == current_hash
            else "MISMATCH"
        )

        print("\nEvidence ID   :", evidence_id)
        print("Recorded Hash :", recorded_hash)
        print("Current Hash  :", current_hash)
        print("Result        :", status)

    print(
        "\nRESULT: Evidence integrity verification completed."
    )


# ============================================================
# 5. BASELINE VALIDATION
# ============================================================

def validate_baseline():

    print("\n" + "=" * 110)
    print("5. BASELINE VS CURRENT SNAPSHOT VALIDATION")
    print("=" * 110)

    results = []

    used_current = set()

    # Files with same filename
    for filename, old_hash in BASELINE.items():

        if filename in CURRENT_SNAPSHOT:

            used_current.add(filename)

            if old_hash == CURRENT_SNAPSHOT[filename]:
                status = "UNCHANGED"
            else:
                status = "MODIFIED"

            results.append(
                (filename, filename, status)
            )

    # Missing baseline files
    missing_files = [
        filename
        for filename in BASELINE
        if filename not in CURRENT_SNAPSHOT
    ]

    # New current files
    new_files = [
        filename
        for filename in CURRENT_SNAPSHOT
        if filename not in BASELINE
    ]

    renamed_old = set()
    renamed_new = set()

    # Rename detection using same hash
    for old_name in missing_files:

        for new_name in new_files:

            if new_name in renamed_new:
                continue

            if (
                BASELINE[old_name]
                ==
                CURRENT_SNAPSHOT[new_name]
            ):

                results.append(
                    (
                        old_name,
                        new_name,
                        "RENAMED"
                    )
                )

                renamed_old.add(old_name)
                renamed_new.add(new_name)

                break

    # Missing
    for filename in missing_files:

        if filename not in renamed_old:

            results.append(
                (
                    filename,
                    "-",
                    "MISSING"
                )
            )

    # New
    for filename in new_files:

        if filename not in renamed_new:

            results.append(
                (
                    "-",
                    filename,
                    "NEW"
                )
            )

    print(
        f"\n{'Baseline File':<35}"
        f"{'Current File':<35}"
        f"{'Status'}"
    )

    print("-" * 110)

    for old_name, new_name, status in results:

        print(
            f"{old_name:<35}"
            f"{new_name:<35}"
            f"{status}"
        )

    print("\nSUSPICIOUS CHANGES")
    print("-" * 80)

    for old_name, new_name, status in results:

        if status != "UNCHANGED":

            print(
                f"{old_name} -> "
                f"{new_name} : "
                f"{status}"
            )

    return results


# ============================================================
# 6. TIMELINE RECONSTRUCTION
# ============================================================

def reconstruct_timeline():

    timeline = []

    # Authentication source
    for timestamp, user, ip, event, host in AUTH_LOGS:

        timeline.append(
            {
                "timestamp": timestamp,
                "source": "AUTH",
                "details":
                    f"{event} | "
                    f"user={user} | "
                    f"IP={ip} | "
                    f"host={host}"
            }
        )

    # File activity source
    for timestamp, host, user, action, filename in FILE_LOGS:

        timeline.append(
            {
                "timestamp": timestamp,
                "source": "FILE",
                "details":
                    f"{action} | "
                    f"{filename} | "
                    f"user={user} | "
                    f"host={host}"
            }
        )

    # Network source
    for timestamp, host, ip, port, size, protocol in NETWORK_LOGS:

        timeline.append(
            {
                "timestamp": timestamp,
                "source": "NETWORK",
                "details":
                    f"{protocol} | "
                    f"{ip}:{port} | "
                    f"bytes={size} | "
                    f"host={host}"
            }
        )

    timeline.sort(
        key=lambda event:
        datetime.strptime(
            event["timestamp"],
            "%Y-%m-%d %H:%M:%S"
        )
    )

    print("\n" + "=" * 125)
    print("6. MULTI-SOURCE INCIDENT TIMELINE")
    print("=" * 125)

    print(
        "\nSources Correlated:"
        "\n1. Authentication Logs"
        "\n2. File Activity Logs"
        "\n3. Network Logs"
    )

    print("\n" + "-" * 125)

    for event in timeline:

        print(
            f"{event['timestamp']:<22}"
            f"[{event['source']:<8}] "
            f"{event['details']}"
        )

    return timeline


# ============================================================
# 7. INCIDENT ANALYSIS
# ============================================================

def analyze_incident():

    timeline = reconstruct_timeline()

    print("\n" + "=" * 110)
    print("7. INCIDENT ANALYSIS / POINT OF COMPROMISE")
    print("=" * 110)

    failed_logins = [
        log
        for log in AUTH_LOGS
        if log[3] == "FAILED_LOGIN"
        and log[2] == "203.0.113.44"
    ]

    suspicious_success = [
        log
        for log in AUTH_LOGS
        if log[3] == "SUCCESSFUL_LOGIN"
        and log[2] == "203.0.113.44"
    ]

    print("\nAuthentication Analysis")
    print("-" * 70)

    print(
        "Failed login attempts from 203.0.113.44:",
        len(failed_logins)
    )

    if suspicious_success:

        compromise = suspicious_success[0]

        print(
            "Successful external login:",
            compromise[0]
        )

        print(
            "Compromised Account      :",
            compromise[1]
        )

        print(
            "Source IP                :",
            compromise[2]
        )

        print(
            "Affected Workstation     :",
            compromise[4]
        )

    suspicious_bytes = sum(
        size
        for _, _, ip, port, size, _
        in NETWORK_LOGS
        if ip == "198.51.100.77"
        and port == 4444
    )

    print("\nNetwork Analysis")
    print("-" * 70)

    print(
        "Suspicious Destination : 198.51.100.77"
    )

    print(
        "Destination Port       : 4444"
    )

    print(
        "Total Bytes Sent       :",
        suspicious_bytes
    )

    print("\nPROBABLE POINT OF COMPROMISE")
    print("-" * 70)

    print(
        "Timestamp : 2026-03-10 23:58:52"
    )

    print(
        "Event     : Successful login after "
        "six failed login attempts"
    )

    print(
        "Account   : jsmith"
    )

    print(
        "Source IP : 203.0.113.44"
    )

    print(
        "System    : WKSTN-14"
    )

    print("\nINVESTIGATOR INTERPRETATION")
    print("-" * 70)

    print(
        "The authentication sequence is consistent "
        "with successful password guessing or "
        "brute-force access."
    )

    print(
        "Subsequent file changes and large outbound "
        "network transfers are consistent with "
        "post-compromise activity."
    )


# ============================================================
# 8. FORENSIC REPORT
# ============================================================

def generate_report():

    if not evidence_register:
        register_evidence()

    if not custody_log:
        load_chain_of_custody()

    print("\n" + "=" * 105)

    print(
        "DIGITAL FORENSIC EXAMINATION REPORT".center(
            105
        )
    )

    print("=" * 105)

    print("Case ID           :", CASE_ID)
    print("Affected System   :", WORKSTATION)
    print("Lead Investigator :", LEAD_INVESTIGATOR)
    print("Incident Period   : 10-11 March 2026")

    print(
        "Report Generated  :",
        datetime.now().strftime(
            "%Y-%m-%d %H:%M:%S"
        )
    )

    print("\nEVIDENCE EXAMINED")
    print("-" * 105)

    for record in evidence_register:

        print(
            record["evidence_id"],
            "-",
            record["source"]
        )

        print(
            "SHA-256:",
            record["sha256"]
        )

    print("\nCHAIN OF CUSTODY SUMMARY")
    print("-" * 105)

    counts = defaultdict(int)

    for entry in custody_log:
        counts[entry["evidence_id"]] += 1

    for evidence_id, count in counts.items():

        print(
            evidence_id,
            ":",
            count,
            "custody event(s)"
        )

    print("\nFACTUAL FINDINGS")
    print("-" * 105)

    print(
        "1. Six failed login attempts were recorded "
        "for user jsmith from 203.0.113.44."
    )

    print(
        "2. A successful login occurred at "
        "23:58:52 from the same external IP."
    )

    print(
        "3. payroll_master.xlsx was MODIFIED."
    )

    print(
        "4. finance_report_q1.pdf was MISSING."
    )

    print(
        "5. svc_update.exe was identified as NEW."
    )

    print(
        "6. security_policy_old.txt was RENAMED "
        "to security_policy_new.txt."
    )

    print(
        "7. Two outbound TCP connections to "
        "198.51.100.77:4444 transferred "
        "25,220,000 bytes."
    )

    print("\nINVESTIGATOR INTERPRETATION")
    print("-" * 105)

    print(
        "The evidence is consistent with an "
        "external compromise of WKSTN-14."
    )

    print(
        "The successful login following repeated "
        "authentication failures is the most "
        "plausible point of compromise."
    )

    print(
        "Subsequent file manipulation and outbound "
        "network activity indicate potential "
        "post-compromise activity and data exfiltration."
    )

    print("\nCONCLUSION")
    print("-" * 105)

    print(
        "The simulated investigation identified "
        "unauthorized access followed by suspicious "
        "file-system activity and outbound data transfer."
    )

    print(
        "\nMandatory Metadata Check: PASS"
    )

    print(
        "All simulated evidence records contain "
        "the required information."
    )

    print("=" * 105)


# ============================================================
# 9. REFLECTION
# ============================================================

def show_reflection():

    print("\n" + "=" * 100)

    print(
        "9. REFLECTION / DESIGN AND LEGAL-SDG RELEVANCE"
    )

    print("=" * 100)

    print("""
The system was designed as a menu-driven Python application to
represent the complete digital forensic evidence lifecycle.

SHA-256 hashing provides a repeatable method for checking evidence
integrity. Chain-of-custody records maintain accountability by
recording the handler, timestamp, action and purpose of each transfer.

Baseline comparison helps identify modified, missing, new and renamed
files. Timeline reconstruction combines authentication, file activity
and network evidence into a single chronological sequence.

The project has legal relevance because digital evidence should remain
traceable, authentic and verifiable. Maintaining integrity hashes,
custody records and investigator information supports a defensible
forensic examination.

The project relates to SDG 9 through secure and resilient digital
infrastructure, SDG 16 through institutional accountability and
trustworthy investigation, and SDG 17 through standardized information
sharing between security and investigation teams.

The main learning outcome was understanding that digital forensics is
not only about identifying an attacker. Evidence preservation,
integrity verification, chronological reconstruction and responsible
reporting are equally important.
""")


# ============================================================
# 10. COMPLETE DEMONSTRATION
# ============================================================

def complete_demo():

    print("\n" + "#" * 105)

    print(
        "COMPLETE END-TO-END FORENSIC DEMONSTRATION".center(
            105
        )
    )

    print("#" * 105)

    register_evidence()

    load_chain_of_custody()

    display_chain_of_custody()

    verify_integrity()

    validate_baseline()

    reconstruct_timeline()

    analyze_incident()

    generate_report()

    print("\n" + "#" * 105)

    print(
        "END-TO-END DEMONSTRATION COMPLETED SUCCESSFULLY".center(
            105
        )
    )

    print("#" * 105)


# ============================================================
# MENU
# ============================================================

def show_menu():

    print("""
==================================================================
       CAMPUS DIGITAL FORENSICS EVIDENCE MANAGEMENT SYSTEM
==================================================================

1. Register Evidence and Generate SHA-256 Hash
2. Add / Load Chain-of-Custody Records
3. Display Chain-of-Custody History
4. Verify Evidence Integrity
5. Validate Baseline vs Current Snapshot
6. Reconstruct Multi-Source Incident Timeline
7. Analyze Incident and Identify Point of Compromise
8. Generate Structured Forensic Report
9. View Reflection and Legal/SDG Relevance
10. Run Complete End-to-End Demonstration

0. Exit

==================================================================
""")


# ============================================================
# MAIN PROGRAM
# ============================================================

def main():

    while True:

        show_menu()

        choice = input(
            "Enter your choice (0-10): "
        ).strip()

        if choice == "1":

            register_evidence()

        elif choice == "2":

            load_chain_of_custody()

        elif choice == "3":

            display_chain_of_custody()

        elif choice == "4":

            verify_integrity()

        elif choice == "5":

            validate_baseline()

        elif choice == "6":

            reconstruct_timeline()

        elif choice == "7":

            analyze_incident()

        elif choice == "8":

            generate_report()

        elif choice == "9":

            show_reflection()

        elif choice == "10":

            complete_demo()

        elif choice == "0":

            print("\nProgram terminated successfully.")
            break

        else:

            print(
                "\nInvalid option. "
                "Please enter a number between 0 and 10."
            )


# ============================================================
# RUN
# ============================================================

main()
```
